```python
import pandas as pd

df= pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.columns 
```

```python
# create column category age group based on the age column
df['age_group'] = pd.cut(df['age'], bins=[0, 18, 35, 50, 65, 100], labels=['teen', 'young_adult', 'adult', 'middle_aged', 'senior'])
df[['age', 'age_group']].head(10)
```

```python
print(df['frequency_of_purchases'].unique())
```

```python
# turning frequency of purchase into numeber of days between purchases, so that we can use it in our analysis

df['frequency_of_purchases_in_days'] = df['frequency_of_purchases'].map({'Fortnightly': 14, 'Monthly': 30, 'Quarterly': 90, 'Weekly': 7, 'Annually': 365, 'Bi-Weekly': 14, 'Every 3 Months': 90 })
```

```python
print(df['frequency_of_purchases_in_days'], df['frequency_of_purchases'])
```

```python
#discount can be applied through promo code or through a discount applied directly to the product. We can check if these two columns are the same or not.
df[['discount_applied','promo_code_used']]
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
# now we know that the two columns are the same, we can drop one of them. We will drop the promo_code_used column.
df.drop(columns=['promo_code_used'], inplace=True)
```
